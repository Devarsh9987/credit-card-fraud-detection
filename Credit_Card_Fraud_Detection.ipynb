## 1. Project Definition

This project analyzes credit card transaction data to understand the difference between genuine and fraudulent transactions.

## 2. Dataset Use Case

The dataset contains credit card transaction information.

- Transaction Time
- V1 to V28 numerical features

Class 0 represents a genuine transaction and Class 1 represents a fraudulent transaction.

The dataset is used to study the distribution of transactions and identify patterns related to fraudulent transactions.

# 7. Data Visualization

Different visualizations are used to understand genuine and fraudulent transactions based on class, amount, time and numerical features.

### 7.1 Genuine vs Fraudulent Transactions

This graph compares the number of genuine and fraudulent transactions in the dataset.

```python
class_counts = df['Class'].value_counts()

plt.figure(figsize=(8, 5))
class_counts.plot(kind='bar')

plt.title('Genuine vs Fraudulent Transactions')
plt.xlabel('Transaction Class')
plt.ylabel('Number of Transactions')
plt.xticks([0, 1], ['Genuine', 'Fraud'], rotation=0)

plt.show()
```

### 7.2 Transaction Amount Distribution

This histogram shows the distribution of transaction amounts in the dataset.

```python
plt.figure(figsize=(10, 6))

plt.hist(df['Amount'], bins=50)

plt.title('Distribution of Transaction Amount')
plt.xlabel('Transaction Amount')
plt.ylabel('Number of Transactions')

plt.show()
```

### 7.3 Transaction Amount: Genuine vs Fraud

This graph compares transaction amounts between genuine and fraudulent transactions.

```python
plt.figure(figsize=(8, 6))

sns.boxplot(x='Class', y='Amount', data=df)

plt.title('Transaction Amount: Genuine vs Fraud')
plt.xlabel('Transaction Class')
plt.ylabel('Transaction Amount')
plt.xticks([0, 1], ['Genuine', 'Fraud'])

plt.show()
```

### 7.4 Transaction Time Distribution

This graph shows how transactions are distributed over the recorded time period.

```python
plt.figure(figsize=(10, 6))

plt.hist(df['Time'], bins=50)

plt.title('Transaction Time Distribution')
plt.xlabel('Time')
plt.ylabel('Number of Transactions')

plt.show()
```

### 7.5 Fraud Transactions Over Time

This graph shows the occurrence of fraudulent transactions across the recorded time period.

```python
fraud_data = df[df['Class'] == 1]

plt.figure(figsize=(10, 6))

plt.scatter(fraud_data['Time'], fraud_data['Amount'], alpha=0.6)

plt.title('Fraud Transactions Over Time')
plt.xlabel('Time')
plt.ylabel('Transaction Amount')

plt.show()
```

### 7.6 Correlation Heatmap

This heatmap shows the correlation between the numerical features in the dataset.

```python
plt.figure(figsize=(8, 5))

sns.heatmap(df[['Time', 'Amount', 'Class']].corr(), annot=True)

plt.title('Correlation Heatmap')
plt.show()
```

### 7.7 Fraud Transaction Amount Distribution

This graph shows the distribution of transaction amounts only for fraudulent transactions.

```python
plt.figure(figsize=(10, 6))

plt.hist(fraud_data['Amount'], bins=30)

plt.title('Fraudulent Transaction Amount Distribution')
plt.xlabel('Transaction Amount')
plt.ylabel('Number of Fraud Transactions')

plt.show()
```

# 8. Key Findings

- Genuine transactions are much more common than fraudulent transactions.
- Fraudulent transactions form only a small part of the dataset.
- Most transactions have relatively small amounts.
- Transaction amounts vary between genuine and fraudulent transactions.
- Fraudulent transactions occur at different points in time.
- The correlation heatmap helps identify relationships between numerical features.
- The dataset shows that credit card fraud is a highly imbalanced problem.
